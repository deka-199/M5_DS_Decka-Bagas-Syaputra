```python
import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()

df = pd.read_csv(next(iter(uploaded)))

print("Ukuran Dataset:", df.shape)
df.head()
```

```python
df.info()
```

```python
# Memeriksa tipe data harga_satuan
print(df['harga_satuan'].dtype)

# Mengganti koma menjadi titik
df['harga_satuan'] = df['harga_satuan'].str.replace(',', '.', regex=False)

# Mengubah tipe data menjadi float
df['harga_satuan'] = df['harga_satuan'].astype(float)

# Memeriksa hasil perubahan
print(df['harga_satuan'].dtype)
df[['harga_satuan']].head(10)
```

```python
# Memeriksa jumlah data negatif
print("Jumlah data negatif:", (df['jumlah'] < 0).sum())

# Mengubah nilai negatif menjadi positif
df.loc[df['jumlah'] < 0, 'jumlah'] = df.loc[df['jumlah'] < 0, 'jumlah'].abs()

# Memeriksa hasil perubahan
print("Jumlah data negatif setelah perbaikan:", (df['jumlah'] < 0).sum())
```

```python
# Memeriksa jumlah nilai kosong
print("Nilai kosong sebelum imputasi:", df['kode_kasir'].isna().sum())

# Mengisi nilai kosong
df.loc[df['kode_kasir'].isna(), 'kode_kasir'] = 'K_UNKNOWN'

# Memeriksa hasil imputasi
print("Nilai kosong setelah imputasi:", df['kode_kasir'].isna().sum())
```

```python
df.iloc[0:5, [0, 1, 2, 5]]
```

```python
# Menghitung omzet kotor
df['Total_Kotor'] = df['harga_satuan'] * df['jumlah']

# Menampilkan hasil perhitungan
df[['harga_satuan', 'jumlah', 'Total_Kotor']].head(10)
```

```python
# Menghitung nominal diskon
df['Nominal_Diskon'] = df['Total_Kotor'] * (df['diskon_persen'] / 100)

# Menampilkan hasil perhitungan
df[['Total_Kotor', 'diskon_persen', 'Nominal_Diskon']].head(10)
```

```python
# Menghitung omzet bersih setelah diskon dan pajak
df['Total_Bersih'] = (df['Total_Kotor'] - df['Nominal_Diskon']) * 1.11

# Menampilkan hasil perhitungan
df[['Total_Kotor', 'Nominal_Diskon', 'Total_Bersih']].head(10)
```

```python
# Memeriksa informasi kolom baru
df[['Total_Kotor', 'Nominal_Diskon', 'Total_Bersih']].describe()
```

```python
# Melakukan filtering berdasarkan beberapa kondisi
filter_transaksi = df.loc[
    (df['status'] == 'SELESAI') &
    (df['kode_gerai'].isin(['G05', 'G09'])) &
    (df['Total_Bersih'] > 100000)
]

# Menampilkan hasil filtering
display(filter_transaksi)

# Menghitung jumlah transaksi yang memenuhi kriteria
print("Jumlah transaksi yang memenuhi kriteria:", len(filter_transaksi))
```

```python
# Menampilkan transaksi gerai G10 dengan jumlah minimal 5 unit
filter_query = df.query("kode_gerai == 'G10' and jumlah >= 5")

# Menampilkan hasil filtering
display(filter_query)

# Menghitung jumlah transaksi
print("Jumlah transaksi gerai G10 dengan minimal 5 unit:", len(filter_query))
```

```python
# Memeriksa jumlah hasil filtering
print("Hasil Filter Multi-Kondisi:", filter_transaksi.shape[0], "baris")
print("Hasil Filter Query:", filter_query.shape[0], "baris")
```

```python
# Mengurutkan data berdasarkan kode gerai dan omzet bersih
df_sorted = df.sort_values(
    by=['kode_gerai', 'Total_Bersih'],
    ascending=[True, False]
)

# Menampilkan hasil pengurutan
display(df_sorted.head(10))
```

```python
# Mengambil lima transaksi dengan omzet bersih tertinggi
top_5_transaksi = df.nlargest(5, 'Total_Bersih')

# Menampilkan lima transaksi tertinggi
display(top_5_transaksi)
```

```python
# Memeriksa urutan lima transaksi tertinggi
print("Lima transaksi dengan Total_Bersih tertinggi:")
print(top_5_transaksi['Total_Bersih'].tolist())
```

```python
# Memfilter transaksi dengan status SELESAI
df_selesai = df.loc[df['status'] == 'SELESAI']

# Membuat laporan agregasi berdasarkan kode gerai
laporan_gerai = df_selesai.groupby('kode_gerai').agg(
    Jumlah_Transaksi=('id_transaksi', 'count'),
    Total_Unit_Terjual=('jumlah', 'sum'),
    Total_Omzet_Bersih=('Total_Bersih', 'sum'),
    Rata_Rata_Transaksi=('Total_Bersih', 'mean')
)

# Mengurutkan laporan berdasarkan omzet bersih tertinggi
laporan_gerai = laporan_gerai.sort_values(
    by='Total_Omzet_Bersih',
    ascending=False
).round(2)

# Menampilkan laporan kinerja gerai
display(laporan_gerai)
```

```python
# Menghitung total omzet bersih setiap gerai
total_omzet_gerai = df.groupby('kode_gerai')['Total_Bersih'].transform('sum')

# Menghitung persentase kontribusi setiap transaksi
df['Kontribusi_Gerai_Persen'] = (df['Total_Bersih'] / total_omzet_gerai) * 100

# Menampilkan hasil transformasi
display(df[['id_transaksi', 'kode_gerai', 'Total_Bersih', 'Kontribusi_Gerai_Persen']].head(10))
```

```python
# Memeriksa total persentase kontribusi setiap gerai
cek_kontribusi = df.groupby('kode_gerai')['Kontribusi_Gerai_Persen'].sum()

display(cek_kontribusi)
```
